```python
import pandas as pd

# Load the Excel file
insurance_df = pd.read_excel("./insurance.xlsx")

# Display the first few rows
insurance_df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, mean_squared_error
import numpy as np

# Create binary target for classification: 1 if charges > median, else 0
insurance_df['high_charges'] = (insurance_df['charges'] > insurance_df['charges'].median()).astype(int)

# Define features and targets
features = ['age', 'sex', 'bmi', 'children', 'smoker', 'region']
X = insurance_df[features]
y_regression = insurance_df['charges']
y_classification = insurance_df['high_charges']

# Train-test split
X_train, X_test, y_reg_train, y_reg_test, y_cls_train, y_cls_test = train_test_split(
    X, y_regression, y_classification, test_size=0.2, random_state=42
)

# Preprocessing: encode categoricals, scale numericals
numeric_features = ['age', 'bmi', 'children']
categorical_features = ['sex', 'smoker', 'region']

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_features),
    ("cat", OneHotEncoder(), categorical_features)
])

# -------------------------
# 1. Linear Regression
# -------------------------
lr_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
lr_pipeline.fit(X_train, y_reg_train)
lr_preds = lr_pipeline.predict(X_test)
lr_mse = mean_squared_error(y_reg_test, lr_preds)

# -------------------------
# 2. SVM Classification
# -------------------------
svm_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", SVC())
])
svm_pipeline.fit(X_train, y_cls_train)
svm_preds = svm_pipeline.predict(X_test)
svm_accuracy = accuracy_score(y_cls_test, svm_preds)

# -------------------------
# 3. Bayesian Classifier
# -------------------------
nb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", GaussianNB())
])
nb_pipeline.fit(X_train, y_cls_train)
nb_preds = nb_pipeline.predict(X_test)
nb_accuracy = accuracy_score(y_cls_test, nb_preds)

# Output results
{
    "Linear Regression MSE": lr_mse,
    "SVM Classification Accuracy": svm_accuracy,
    "Bayesian Classifier Accuracy": nb_accuracy
}
```
